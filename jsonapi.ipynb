```python
from datetime import datetime
from typing import List
from pydantic import BaseModel, Field
from openai import OpenAI
import json


class ProductoSEO(BaseModel):
    nombre: str = Field(description="Nombre comercial y atractivo del producto.")
    descripcion: str = Field(description="Descripción optimizada para SEO, persuasiva y orientada a la conversión.")
    palabras_clave: List[str] = Field(description="Lista de 5 palabras clave long-tail recomendadas para posicionar el producto.")
    puntuacion_seo_estimada: int = Field(description="Puntuación estimada de optimización SEO del 1 al 100.")

class MetaDatos(BaseModel):
    fecha_generacion: str = Field(default_factory=lambda: datetime.now().strftime("%Y-%m-%d %H:%M:%S"), description="Fecha y hora exacta de la generación de la ficha.")
    modelo_utilizado: str = Field(description="Nombre del modelo de IA que procesó la solicitud.")

class FichaProductoResponse(BaseModel):
    producto: ProductoSEO
    metadatos: MetaDatos



FAKE_API_KEY = "sk-YOUR_FAKE_API_KEY"

client = OpenAI(
    api_key=FAKE_API_KEY
)


producto_entrada = {
    "nombre_base": "Chronos Smartwatch V2",
    "caracteristicas": "Batería de 14 días, monitor de oxígeno en sangre, pantalla AMOLED, sumergible a 50 metros, color negro mate."
}


prompt_sistema = (
    "Eres un especialista senior en SEO y Copywriting para comercio electrónico. "
    "Tu objetivo es transformar especificaciones técnicas crudas en fichas de producto JSON optimizadas "
    "que sigan extrictamente la estructura estructurada provista."
)

prompt_usuario = (
    f"Por favor, genera la optimización SEO para el siguiente producto:\n"
    f"Producto: {producto_entrada['nombre_base']}\n"
    f"Características técnicas: {producto_entrada['caracteristicas']}"
)

print("Iniciando petición simulada a OpenAI utilizando respuestas estructuradas...")

try:

    
    completion = client.beta.chat.completions.parse(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": prompt_sistema},
            {"role": "user", "content": prompt_usuario},
        ],
        response_format=FichaProductoResponse,
    )
    

    resultado_validado = completion.choices[0].message.parsed
    print("\n[ÉXITO] Respuesta recibida y validada correctamente por Pydantic:")
    print(json.dumps(resultado_validado.model_dump(), indent=4, ensure_ascii=False))

except Exception as e:
    print(f"\n[AVISO DE SIMULACIÓN] La llamada falló o fue interceptada de forma esperada.")
    print(f"Detalle del error (ej. Clave de API no válida): {e}")
    

    print("\n--- Ejecutando simulación local de validación con JSON de prueba ---")
    
    json_simulado_correcto = {
        "producto": {
            "nombre": "Smartwatch Chronos V2 – El Reloj Inteligente con Batería de 14 Días",
            "descripcion": "Lleva tu rendimiento al siguiente nivel con el nuevo Chronos Smartwatch V2. Diseñado con una espectacular pantalla AMOLED de alta definición y acabado negro mate, este reloj inteligente es el aliado perfecto para tu día a día. Olvídate de los cargadores gracias a su batería de larga duración de hasta 14 días. Monitoriza tu salud en tiempo real con su avanzado sensor de oxígeno en sangre y entrena sin límites bajo el agua gracias a su resistencia sumergible hasta 50 metros. Compra hoy el Chronos V2 y domina tu tiempo.",
            "palabras_clave": [
                "smartwatch bateria 14 dias",
                "reloj inteligente pantalla amoled",
                "reloj sumergible 50 metros",
                "smartwatch con oxigeno en sangre",
                "Chronos Smartwatch V2 opiniones"
            ],
            "puntuacion_seo_estimada": 95
        },
        "metadatos": {
            "fecha_generacion": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            "modelo_utilizado": "gpt-4o-mini"
        }
    }
    
    try:

        objeto_validado = FichaProductoResponse(**json_simulado_correcto)
        print("\n[ÉXITO LOCAL] El JSON simulado cumple estrictamente con el esquema Pydantic.")
        print(f"Producto Validado: {objeto_validado.producto.nombre}")
        print(f"Keywords detectadas: {', '.join(objeto_validado.producto.palabras_clave)}")
        print(f"Puntuación SEO: {objeto_validado.producto.puntuacion_seo_estimada}/100")
        
    except Exception as validation_error:
        print(f"\n[ERROR DE VALIDACIÓN] El JSON devuelto no cumple el esquema establecido: {validation_error}")
```
